```python
from pyspark.sql import functions as F

dbutils.widgets.text("pipeline_name", "")
dbutils.widgets.text("adf_run_id", "")
dbutils.widgets.text("child_run_id", "")
dbutils.widgets.text("status", "")
dbutils.widgets.text("error_message", "")
dbutils.widgets.text("trigger_name", "")
dbutils.widgets.text("trigger_time", "")

AUDIT = "olist_dev.silver.pipeline_audit"

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {AUDIT} (
  pipeline_name STRING, adf_run_id STRING, child_run_id STRING,
  status STRING, error_message STRING,
  trigger_name STRING, trigger_time STRING, logged_ts TIMESTAMP)
""")

row = [(dbutils.widgets.get("pipeline_name"),
        dbutils.widgets.get("adf_run_id"),
        dbutils.widgets.get("child_run_id"),
        dbutils.widgets.get("status"),
        dbutils.widgets.get("error_message")[:2000],
        dbutils.widgets.get("trigger_name"),
        dbutils.widgets.get("trigger_time"))]

(spark.createDataFrame(row,
    "pipeline_name string, adf_run_id string, child_run_id string, status string, "
    "error_message string, trigger_name string, trigger_time string")
 .withColumn("logged_ts", F.current_timestamp())
 .write.mode("append").saveAsTable(AUDIT))

dbutils.notebook.exit("logged")
```
